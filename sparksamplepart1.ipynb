```python

cus_df = spark.sql("select customerID, first_name, country from samples.bakehouse.sales_customers")
cus_df.display()
     
```

```python
from pyspark.sql import functions as F
```

```python
result_df=(
    
    cus_df
    .select(        
        F.col ("customerID"),
        F.col("first_name"),
        F.col("country"),
        F.col("gender") ,
        F.col("email_address") 
    )
)
result_df.display()

```

```python
result_df=(
    cus_df
    .filter(F.col("gender")=='female')
    
    
)
result_df.display()
```

```python
result_df=(
    cus_df
    .filter(
        (F.col("gender")=='male')&
         (F.col("country")=='USA')

    )
    
 
)
result_df.display()
```

```python
result_df=(
    cus_df
    .where(
        (F.col("gender")=='male')&
         (F.col("country")=='USA')

    )
    
 
)
result_df.display()
```
