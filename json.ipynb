```python
cus_df=(
 spark.read.option("multiLine", True).json("/Volumes/rg_qc_batch10/swapna/swapna/elements.json")
)
cus_df.display()
```

```python
cus_df = (
    cus_df.withColumn("orders", F.explode(F.col("elements")))
    .select(
        F.col("orders.atomic_mass").alias("atomic_mass"),
        F.col("orders.category").alias("category"),
        F.col("orders.discovered").alias("discovered"),
        F.col("orders.name").alias("name"),
        F.col("orders.number").alias("number"),
        F.col("orders.phase").alias("phase"),
        F.col("orders.symbol").alias("symbol"),
    )
)
cus_df.display()
```
