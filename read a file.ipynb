```python
from pyspark.sql import functions as F
from pyspark.sql import Window
```

```python
cus_df=(
 spark.read.option("multiLine", True).json("/Volumes/rg_qc_batch10/swapna/swapna/currencies.json")
)
cus_df.display()
```

```python
cus_df=(
    cus_df.withColumn("currencies",F.explode(F.col("currencies")))
    .select(
        F.col("currencies.code").alias("code_id"),
        F.col("currencies.countries").alias("country"),
        F.col("currencies.decimal_digits").alias("digital_id"),
        F.col("currencies.name").alias("name_id"),
        F.col("currencies.symbol").alias("symbol")
    )
)
cus_df.display()
```

```python
adls_path= "abfss://adbdfcontainer@qcadlsbatch.dfs.core.windows.net/"
```

```python
target_path = f"{adls_path}/curated_data/" + 'currencies.json'
cus_df.write.format("json").mode("overwrite").option("header", True).save(target_path)
```
