```python

from pyspark.sql import functions as F
```

```python
adls_path= "abfss://adbdfcontainer@qcadlsbatch.dfs.core.windows.net/"
file_path=f"{adls_path}/raw-container/selected-price-indexes-august-2026.csv"

price_df=(
    spark.read.format("csv")
    .option("header","true")
    .option("inferschema","true")
    .load(file_path)

)
price_df.display()
```

```python
price_df=(
    price_df
    .withColumn("period",F.col("Period").cast("int"))
    .withColumn("data_value",F.col("data_value").try_cast("double"))
    .withColumn("data_value",F.col("data_value").try_cast("int"))
    .groupBy("period")
    .agg(F.sum("Data_value").alias("total_data_value"))

)
price_df.display()
```

```python
target_path = f"{adls_path}/curated_data/selected_price_indexes_total"
price_df.write.format("csv").mode("overwrite").option("header", True).save(target_path)
```

```python
target_path=f"{adls_path}/curtated_data/selected_price_indexes_total_parquet"
price_df.write.format("parquet").mode("overwrite").save(target_path)
```

```python
p_df = spark.read.format("parquet").load(f"{adls_path}/curtated_data/selected_price_indexes_total_parquet")

p_df.display()
```
