```python
%sql
SELECT * FROM `samples`.`bakehouse`.`sales_transactions`;
```

```python
sales_df = (
    spark.table("samples.bakehouse.sales_transactions")
)

sales_df.display()
```

```python
%sql
select
customerID,
sum (quantity) as total_quantity,
count(*) as total_transactions
from samples.bakehouse.sales_transactions
group by customerID;
```

```python
result_df=(
    sales_df
.groupBy("customerID")
.count()
)
result_df.display()

    
```

```python
result_df=(
    sales_df
    .groupBy("customerID")
    .agg(
        F.sum("quantity"). alias("total_quantity"),
        F.count("*").alias("total_transcations"),
        F.max("quantity").alias("max_quantity"),
        F.min("quantity").alias("min_quantity")

       
    )
)
result_df.display()
```

```python
%sql
select
customerID,
product,
sum(quantity)as total_quantity
from samples.bakehouse.sales_transactions
group by customerID,product;
```

```python
%sql
select
    customerID,
    product,
    sum(quantity) as total_quantity
from samples.bakehouse.sales_transactions
group by customerID, product
having sum(quantity) > 50;
```

```python
result_df=(
    sales_df
   . groupBy("customerID","product")
   .agg(
       F.sum("quantity").alias("total_quantity")
   )
   .filter(F.col("total_quantity") > 50)
)
result_df.display()
```
